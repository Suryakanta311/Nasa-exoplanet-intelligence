```python
import lightkurve as lk
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
print("all libraries are imported succesfully")
```

```python
search_result=lk.search_lightcurve(
    "kepler-10",
     mission="kepler"
)
print(search_result)
```

```python
#download lightcurve for a observation
lightcurve=search_result[0].download()
print(lightcurve)
```

```python
# Check the different quality flag values
print(lightcurve.to_table().to_pandas()["quality"].value_counts())
```

```python
good=(lightcurve.quality==0).sum()
flaged=(lightcurve.quality!=0).sum()
print("good",good)
print("flaged",flaged)
print("total measurement",len(lightcurve))
```
